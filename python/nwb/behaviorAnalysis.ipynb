```python
dataDir = Path(r'Z:\sueSu\tempTest\447')
rigs = ['1', '2', '3']
boxs = ['A', 'B', 'C', 'D']
```

```python
for rCurr, rig in enumerate(rigs):
    for bCurr, box in enumerate(boxs):
        currFolder = f'{dataDir}\\{rig}{box}'
        sessionFile = []
        for root, dirs, files in os.walk(currFolder):
            for file in files:
                if file.endswith(".json"):
                    sessionFile.append(os.path.join(root, file))

        json_folder = sessionFile[-1]
        behData = json.load(open(json_folder))
        leftLicks = behData['B_LeftLickTime']
        rightLicks = behData['B_RightLickTime']
        allLicks = np.sort(np.concatenate((rightLicks, leftLicks)))
        sortInd = np.argsort(np.concatenate((rightLicks, leftLicks)))
        allLicksID = np.concatenate((np.ones_like(rightLicks), np.zeros_like(leftLicks)))
        allLicksID = allLicksID[sortInd]
        allLickDiffs = np.diff(allLicks)

        fig, axs = plt.subplots(1, 5, figsize=(16, 4))
        axs[0].hist(1000*allLickDiffs[(allLicksID[:-1]==0) & (allLicksID[1:]==0)], bins=np.arange(0, 500, 20))
        axs[0].set_title('L-L')
        axs[0].set_xlabel('ms')
        axs[1].hist(1000*allLickDiffs[(allLicksID[:-1]==1) & (allLicksID[1:]==1)], bins=np.arange(0, 500, 20))
        axs[1].set_title('R-R')
        axs[1].set_xlabel('ms')
        axs[2].hist(1000*allLickDiffs[(allLicksID[:-1]==1) & (allLicksID[1:]==0)], bins=np.arange(0, 500, 20))
        axs[2].set_title('R-L')
        axs[2].set_xlabel('ms')
        axs[3].hist(1000*allLickDiffs[(allLicksID[:-1]==0) & (allLicksID[1:]==1)], bins=np.arange(0, 500, 20))
        axs[3].set_title('L-R')
        axs[3].set_xlabel('ms')
        axs[4].hist(1000*allLickDiffs, bins=np.arange(0, 500, 20))
        axs[4].set_title('All')
        axs[4].set_xlabel('ms')

        plt.suptitle(f'{rig}{box}')
        saveName = f'{dataDir}\\{rig}{box}iliDist2.pdf'
        plt.savefig(saveName)  
        plt.close()
```
